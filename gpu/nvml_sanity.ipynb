Run on Colab (GPU runtime) or the lab RTX 5090. Confirms `pynvml` can read GPU power before wiring up the full benchmark.

```python
import pynvml, time
pynvml.nvmlInit()
handle = pynvml.nvmlDeviceGetHandleByIndex(0)
name = pynvml.nvmlDeviceGetName(handle)
print('GPU:', name)

for _ in range(5):
    p = pynvml.nvmlDeviceGetPowerUsage(handle) / 1000
    print(f'{p:.2f} W')
    time.sleep(0.1)
```
